```python
import sqlite3
DB_5 = "db_5.db"

conn = None
try:
    conn = sqlite3.connect(DB_5)
    cursor = conn.cursor()

    cursor.execute("CREATE TABLE IF NOT EXISTS SanPham (ID INTEGER PRIMARY KEY, TenSP TEXT, Gia REAL);")
    
    products = [
        (1, 'Sách Lập Trình Python', 35000.0),
        (2, 'Bút Bi', 5000.0),
        (3, 'Máy Tính Cầm Tay', 250000.0)
    ]
    cursor.executemany("INSERT OR REPLACE INTO SanPham VALUES (?, ?, ?)", products)
    conn.commit()
    print(f"➕ Đã chèn {cursor.rowcount} bản ghi vào bảng SanPham.")
    cursor.execute("SELECT ID, TenSP, Gia FROM SanPham")
    
    print("\n📊 Dữ liệu trong bảng SanPham:")
    print(f"{'ID':<4} | {'Tên Sản Phẩm':<20} | {'Giá (VNĐ)':<10}")
    print("-" * 38)
    for row in cursor.fetchall():
        print(f"{row[0]:<4} | {row[1]:<20} | {row[2]:<10.0f}")
        
except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
```
