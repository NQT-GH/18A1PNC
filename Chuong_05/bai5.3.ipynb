```python
import sqlite3
DB_3 = "db_3.db"

conn = None
try:
    conn = sqlite3.connect(DB_3)
    cursor = conn.cursor()
    create_table_sql = """
    CREATE TABLE IF NOT EXISTS KhachHang (
        MaKH INTEGER PRIMARY KEY,
        TenKH TEXT NOT NULL,
        Email TEXT UNIQUE
    );
    """
    cursor.execute(create_table_sql)
    conn.commit()
    
    print(f"✅ Đã tạo bảng 'KhachHang' trong {DB_3}")
    
except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
```
