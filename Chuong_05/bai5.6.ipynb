```python
import sqlite3
DB_5 = "db_5.db"

conn = None
try:
    conn = sqlite3.connect(DB_5)
    cursor = conn.cursor()
    cursor.execute("SELECT COUNT(*) FROM SanPham")
    count = cursor.fetchone()[0]
    
    print(f"🔢 Tổng số bản ghi trong bảng SanPham là: {count}")
    
except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
```
