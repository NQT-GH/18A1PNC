```python
import sqlite3
DB_5 = "db_5.db"

conn = None
try:
    conn = sqlite3.connect(DB_5)
    cursor = conn.cursor()
    
    id_can_xoa = 2
    
    cursor.execute("DELETE FROM SanPham WHERE ID = ?", (id_can_xoa,))
    conn.commit()
    
    print(f"🗑️ Đã xóa thành công {cursor.rowcount} hàng có ID = {id_can_xoa}.")
    
    cursor.execute("SELECT ID, TenSP FROM SanPham")
    rows_con_lai = cursor.fetchall()
    
    print("\n📝 Danh sách Sản Phẩm còn lại:")
    if rows_con_lai:
        for row in rows_con_lai:
            print(f"- ID: {row[0]}, Tên: {row[1]}")
    else:
        print("Bảng đã rỗng.")

except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
```
