```python
import sqlite3
DB_1 = "db_1.db"

try:
    conn = sqlite3.connect(DB_1)
    cursor = conn.cursor()
    cursor.execute("SELECT sqlite_version();")
    version = cursor.fetchone()
    print(f"✅ Đã kết nối thành công đến {DB_1}")
    print(f"📦 Phiên bản SQLite đang sử dụng: {version[0]}")
    
except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
        print("Đã đóng kết nối.")
```
