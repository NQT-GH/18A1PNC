```python
import sqlite3
conn = None

try:
    conn = sqlite3.connect(':memory:')
    print("✅ Đã kết nối thành công đến CSDL")
    cursor = conn.cursor()
    cursor.execute("CREATE TABLE TempData (value INTEGER);")
    cursor.execute("INSERT INTO TempData VALUES (100);")
    conn.commit()
    print("💡 Đã tạo và chèn dữ liệu tạm thời.")
    
except sqlite3.Error as e:
    print(f"❌ Lỗi xảy ra: {e}")
finally:
    if conn:
        conn.close()
        print("Đã đóng kết nối")
```
